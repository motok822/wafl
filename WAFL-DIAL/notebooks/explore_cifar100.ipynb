# CIFAR-100 データセット探索

このノートブックでCIFAR-100のデータを確認できます

```python
import matplotlib.pyplot as plt
import numpy as np
import torchvision
from torchvision import transforms
```

```python
# データセットの読み込み
transform = transforms.Compose([transforms.ToTensor()])

train_dataset = torchvision.datasets.CIFAR100(
    root="./data", train=True, download=True, transform=transform
)
test_dataset = torchvision.datasets.CIFAR100(
    root="./data", train=False, download=True, transform=transform
)
```

```python
# CIFAR-100のクラス名（fine labels）
fine_labels = [
    "apple", "aquarium_fish", "baby", "bear", "beaver", "bed", "bee", "beetle",
    "bicycle", "bottle", "bowl", "boy", "bridge", "bus", "butterfly", "camel",
    "can", "castle", "caterpillar", "cattle", "chair", "chimpanzee", "clock",
    "cloud", "cockroach", "couch", "crab", "crocodile", "cup", "dinosaur",
    "dolphin", "elephant", "flatfish", "forest", "fox", "girl", "hamster",
    "house", "kangaroo", "keyboard", "lamp", "lawn_mower", "leopard", "lion",
    "lizard", "lobster", "man", "maple_tree", "motorcycle", "mountain", "mouse",
    "mushroom", "oak_tree", "orange", "orchid", "otter", "palm_tree", "pear",
    "pickup_truck", "pine_tree", "plain", "plate", "poppy", "porcupine",
    "possum", "rabbit", "raccoon", "ray", "road", "rocket", "rose", "sea",
    "seal", "shark", "shrew", "skunk", "skyscraper", "snail", "snake", "spider",
    "squirrel", "streetcar", "sunflower", "sweet_pepper", "table", "tank",
    "telephone", "television", "tiger", "tractor", "train", "trout", "tulip",
    "turtle", "wardrobe", "whale", "willow_tree", "wolf", "woman", "worm"
]
```

```python
# データセットの基本情報
print(f"訓練データ数: {len(train_dataset)}")
print(f"テストデータ数: {len(test_dataset)}")
print(f"クラス数: {len(fine_labels)}")
print(f"画像サイズ: {train_dataset[0][0].shape}")
```

```python
# クラス分布を確認
class_counts = {}
for _, label in train_dataset:
    if label not in class_counts:
        class_counts[label] = 0
    class_counts[label] += 1

print(f"各クラスのサンプル数: {class_counts[0]} (均等分布)")
print(f"全クラス数: {len(class_counts)}")
```

```python
# ランダムなサンプル画像を表示
fig, axes = plt.subplots(4, 8, figsize=(16, 8))
fig.suptitle("CIFAR-100 ランダムサンプル", fontsize=16)

for i in range(32):
    idx = np.random.randint(0, len(train_dataset))
    image, label = train_dataset[idx]
    
    # Tensor to numpy (C, H, W) -> (H, W, C)
    image_np = image.numpy().transpose(1, 2, 0)
    
    ax = axes[i // 8, i % 8]
    ax.imshow(image_np)
    ax.set_title(f"{fine_labels[label]}", fontsize=8)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
# 特定のクラスの画像を表示する関数
def show_class_samples(class_name, num_samples=16):
    """
    指定したクラスのサンプル画像を表示
    
    Args:
        class_name: クラス名 (例: 'bear', 'apple', 'lion')
        num_samples: 表示するサンプル数
    """
    if class_name not in fine_labels:
        print(f"エラー: '{class_name}' は有効なクラス名ではありません")
        print(f"有効なクラス名: {fine_labels}")
        return
    
    target_label = fine_labels.index(class_name)
    
    images = []
    for image, label in train_dataset:
        if label == target_label:
            images.append(image)
            if len(images) >= num_samples:
                break
    
    # グリッドサイズを計算
    cols = 4
    rows = (num_samples + cols - 1) // cols
    
    fig, axes = plt.subplots(rows, cols, figsize=(12, 3 * rows))
    fig.suptitle(f"CIFAR-100: '{class_name}' クラスのサンプル", fontsize=16)
    
    if rows == 1:
        axes = [axes]
    
    for i in range(num_samples):
        if i >= len(images):
            break
        image_np = images[i].numpy().transpose(1, 2, 0)
        ax = axes[i // cols][i % cols] if rows > 1 else axes[i % cols]
        ax.imshow(image_np)
        ax.axis("off")
    
    # 余った枠を非表示
    for i in range(num_samples, rows * cols):
        ax = axes[i // cols][i % cols] if rows > 1 else axes[i % cols]
        ax.axis("off")
    
    plt.tight_layout()
    plt.show()
```

```python
# 例: bearクラスのサンプルを表示
show_class_samples('bear', num_samples=16)
```

```python
# 例: 他のクラスも試してみる
show_class_samples('apple', num_samples=12)
```
